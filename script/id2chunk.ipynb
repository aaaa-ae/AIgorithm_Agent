```python
import json
import re
import os

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
txt_path = os.path.join(PROJECT_ROOT, "data", "knowledge_graph", "test.txt") #输入文件路径
json_path = os.path.join(PROJECT_ROOT, "data", "knowledge_graph", "id2chunk.json") #输出文件路径

with open(txt_path, "r", encoding="utf-8", errors="ignore") as f:
    raw_text = f.read()

# ✅ 核心正则：匹配  id: XXXX  到 下一个 id 或文件结尾
pattern = re.compile(
    r"id:\s*([^\t\n]+)\s*Chunk:\s*(.*?)(?=\nid:\s*[^\t\n]+\s*Chunk:|\Z)",
    re.S
)

matches = pattern.findall(raw_text)

id2chunk = {}

# ✅ 长度统计用的计数器
cnt_0_100 = 0
cnt_100_500 = 0
cnt_500_1000 = 0
cnt_1000_2000 = 0
cnt_2000_plus = 0

for chunk_id, chunk_text in matches:
    clean_text = chunk_text.strip().replace("\n", " ")
    clean_text = re.sub(r"\s+", " ", clean_text)  # 多空格压缩

    cid = chunk_id.strip()
    id2chunk[cid] = clean_text

    L = len(clean_text)

    # ✅ 分段统计
    if L < 100:
        cnt_0_100 += 1
    elif 100 <= L < 500:
        cnt_100_500 += 1
    elif 500 <= L < 1000:
        cnt_500_1000 += 1
    elif 1000 <= L < 2000:
        cnt_1000_2000 +=1
    else:
        cnt_2000_plus += 1

print("✅ 解析出的 chunk 总数:", len(id2chunk))

# ✅ 保存为 JSON（你原来的逻辑不变）
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(id2chunk, f, ensure_ascii=False, indent=2)

print("✅ 已保存为:", json_path)

# =========================
# ✅ 长度区间分布结果
# =========================
print("\n📊 Chunk 长度分布：")
print(f"  < 100 字        : {cnt_0_100}")
print(f"  100 - 500 字    : {cnt_100_500}")
print(f"  500 - 1000 字   : {cnt_500_1000}")
print(f"  1000 - 2000 字   : {cnt_1000_2000}")
print(f"  >= 2000 字      : {cnt_2000_plus}")
```
